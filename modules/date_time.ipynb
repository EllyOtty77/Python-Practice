```python
from datetime import date, datetime, timedelta
```

```python
# Current date
today = date.today()
print(f"Today's date is {today}")
```

```python
# Current date and time
now = datetime.now()
print(f"It's currently {now}")
```

```python
# Creating specific date and time
spec = datetime(2024, 12, 24, 12, 22, 11)
print(spec)
```

```python
# Extracting date and time components
print(f"Year: {now.year}")
print(f"Month: {now.month}")
print(f"Day: {now.day}")
print(f"Hour: {now.hour}")
print(f"Minute: {now.minute}")
print(f"Second: {now.second}")
```

```python
# String formatting
f_date = now.strftime("%A %Y-%m-%d %H:%M")
print(f_date)
```

```python
# Strings to date objects
date_str = "2025-04-12 12:45"
parsed_date = datetime.strptime(date_str, "%Y-%m-%d %H:%M")
print(parsed_date)
```

```python
# Date arithmetic
future = now + timedelta(minutes=50)
print(future)
```

```python
# Timezones
import pytz
```

```python
# UTC timezone
utc_now = datetime.now(pytz.utc)
print(utc_now)
```

```python
# Specific timezone
ny_tz = pytz.timezone("America/New_York")
ny_time = utc_now.astimezone(ny_tz)
print(f"New York Time: {ny_time}")
```
