### Local Scope

```python
# Variables inside a function exist only there:
def greet():
    msg = "hello"   # local scope
    print(msg)

greet()       # hello
# print(msg)  # ❌ NameError: msg is not defined
```

### GLobal Scope

```python
# Declared at the top-level, visible everywhere:
name = "Elly"   # global

def show():
    print("Hi", name)  # can access global

show()    # Hi Elly
print(name)  # Elly
```

```python
# Use the global keyword to modify a global variable inside a function:
count = 0

def increase():
    global count
    count += 1  # modifies global count

increase()
increase()
print(count)   

# Without global, Python would think count is a new local variable.
```

### NonLocal Scope

```python
def outer():
    x = "outer"

    def inner():
        print(x)  # can see enclosing variable
    inner()

outer()  # outer
```

```python
# Use the nononlocal Keyword modify variables in an enclosing function (but not global):
def outer():
    x = 0
    def inner():
        nonlocal x
        x += 1
        return x
    print(inner())  # 1
    print(inner())  # 2

outer()
```
