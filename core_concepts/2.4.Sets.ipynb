# **Sets**

- Sets are unordered collections of unique elements
- Sets are extremely useful when you need to look something up in a large collection, such as filenames, email addresses, or postal codes, because searching is O(1), just as in a dict
- They aid to remove duplicate values from an input list—such as IP addresses in a log file, or the license plate numbers of vehicles that have passed through a parking garage entrance in a given day.

```python
# Sets can be created using {} or set() function
my_set = {1, 2, 3, 5}
my_set
```

```python
set_2 = set([1,2,5,7])
set_2
```

```python
# since sets are unordered, you cannot access elements using index; use in instead
3 in my_set
```

## **Set methods**

```python
# You can add items with  add method
my_set.add(4)
my_set
```

```python
# To add items from another set into the current set, use the update() method.
set1 = {"flex", "antagonist", "self-titled"}
set2 = {"agc", "ns", "broken hearts"}
set2.update(set1)
print(set2)
```

```python
more_songs =  ["chanel", "kills"]
set2.update(more_songs)
print(set2)
```

```python
# use remove or discard to remove items
my_set.remove(2)
my_set
```

```python
set2.discard("chanel")
print(set2)
```

```python
# Remove random item
popped = set2.pop()    
print(popped)
```

```python
# Clear the set
set1.clear()
print(set1)
```

```python
# The del keyword will delete the set completely:
del set1
print(set1)
```

```python
# issubset and issuperset methods
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}

print(set_a.issubset(set_b)) # Returns True if set_a is a subset of set_b
print(set_b.issuperset(set_a)) # Returns True if set_b is a superset of set_a
```

```python
# isdisjoint method
# Returns True if two sets have no elements in common
set_a = {1, 2, 3}
set_b = {4, 5, 6}
print(set_a.isdisjoint(set_b))  # Returns True since there are no common elements 
```

## Join sets

There are several ways to join two or more sets in Python.

- The union() and update() methods joins all items from both sets.
- The intersection() method keeps ONLY the duplicates.
- The difference() method keeps the items from the first set that are not in the other set(s).
- The symmetric_difference() method keeps all items EXCEPT the duplicates.

```python
# Set union
union_set = my_set| set_2
union_set
```

- The  | operator only allows you to join sets with sets, and not with other data types like you can with the  union() method.
- Both union() and update() will exclude any duplicate items.

```python
# Join set and tuple
x = {"a", "b", "c"}
y = (1, 2, 3)

z = x.union(y)
print(z)
```

```python
new_set = my_set.union(set_2)
new_set
```

```python
# Intersection
# Returns a new set with elements common to the set and all others.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.intersection(set2)
print(set3)
```

```python
# You can use the & operator instead of the intersection() method
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 & set2
print(set3)
```

```python
# The intersection_update() method will also keep ONLY the duplicates,
#  but it will change the original set instead of returning a new set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.intersection_update(set2)

print(set1)
```

```python
# Difference
# Returns a new set with elements in the first set that are not in the second set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.difference(set2)
set4 = set1 - set2 # You can also use the - operator

print(set3)
```

```python
# Difference Update
# The difference_update() method will remove the items that exist in both sets from the first set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set1.difference_update(set2)
print(set1)
```

```python
# Symmetric Difference
# Returns a new set with items in either the first or second set, but not both.
set_1 = {"apple", "banana", "cherry"}
set_2 = {"google", "microsoft", "apple"}

sym_diff = set_1.symmetric_difference(set_2)
sym_diff2 = set_1 ^ set_2  # You can also use the ^ operator
print(sym_diff)
```
