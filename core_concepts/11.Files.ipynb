### Read

```python
file = open('cont.txt')
print(file.readline())
```

```python
file.read()
```

```python
# Close the file
file.close()
```

```python
# The file automatically closes
file_path = "cont.txt"
with open(file_path) as file_object:
    for line in file_object:
        print(line.strip().upper())
```

```python
file = open('cont.txt')
lines = file.readlines()
file.close()
lines.sort()
print(lines)
```

### CSV Files

```python
import csv
```

```python
# Read csv file
f = open('demo.csv')
csv_file = csv.reader(f)
for row in csv_file:
    name, age, city = row
    print(f'Name: {name}, Age:{age}, City:{city}')
f.close()
```

```python
# Write to csv
anime = [['AOT', 8.5], ['JJK', 8.4], ['Demon Slayer', 8.4], ['Death Note', 9.0]]

with open('anime.csv', 'w') as acsv: 
    writer = csv.writer(acsv)
    writer.writerows(anime)
```

```python
# Dict reader
with open('anime.csv', 'r') as csv_file:
    reader = csv.DictReader(csv_file)
    for row in reader: 
        print(f'{row["name"]} - {row["rating"]}')
```

```python
# Dict Writer
players = [{'name':'Neymar', 'position':'LW', 'Ovr':94},
           {'name':'Messi', 'position':'RW', 'Ovr':96},
           {'name':'Suker', 'position':'ST', 'Ovr':84},
           {'name':'Mbuemo', 'position':'RW', 'Ovr':74}]
keys = ['name', 'position', 'Ovr']
with open('players.csv', 'w') as write:
    writer = csv.DictWriter(write, fieldnames = keys)
    writer.writeheader()
    writer.writerows(players)
```

### JSON Files

```python
# dump
import json

numbers = [2, 3, 5, 7, 11, 13]
filename = 'numbers.json'
with open(filename, 'w') as f:
    json.dump(numbers, f)

# load
import json

filename = 'numbers.json'
with open(filename) as f:
    numbers = json.load(f)

print(numbers)
```

### Word count

```python
def count_words(filename):
    """Count the approximate number of words in a file."""
    try:
        with open(filename, encoding='utf-8') as f:
            contents = f.read()
    except FileNotFoundError:
        print(f"Sorry, the file {filename} does not exist.")
    else:
        words = contents.split()
        num_words = len(words)
        print(f"The file {filename} has about {num_words} words.")

filename = 'written.txt'
count_words(filename)
```
